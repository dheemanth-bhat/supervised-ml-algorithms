# Random Forest for Classification

## 1 Introduction

### 1.1 Ensembling

```
            Entire Training Dataset
                       │
      ┌────────────────┼────────────────┐
      ↓                ↓                ↓
  Sample #1        Sample #2        Sample #3
      ↓                ↓                ↓
Decision Tree    Decision Tree    Decision Tree
      │                │                │
      └────────────────┼────────────────┘
                       ↓
                  Aggregation
                       ↓
               Final Prediction
```

Random Forest is a special type of Bagging technique where every base-learner in the swarm is a Decision Tree.

#### Properties

- In Random Forest, each of the trees are **Decorrelated** or independent.
- Random Forest is the most popular and proven effective type of Bagging.

### 1.2 Components of Ensembling

#### 1 Base Learner

Random Forest is a special type of Bagging technique where every base learner in the swarm is a Decision Tree.

#### 2 Diversity Mechanism

- Diversity is achieved using Bootstrapping.
- Bootstrapping is sampling with replacement of different subset of features.

#### 3 Aggregation

- For Classification Tasks - The model uses majority voting (or the mode).
- For Regression Tasks - The model uses averaging (or mean).

### 1.3 OOB

#### Properties

- Out-Of-Bag (OOB) data is the remaining samples which are not selected by bootstrapping.
- In Random Forests, OOB data plays a critical role in estimating the model's performance without needing a separate validation set.
- The OOB score in Random Forest models offers a unique method of validation by utilizing unsampled data points during the training process as a test set.

### 1.4 Decorrelation

#### Properties

- In Random Forest, each of the trees are **Decorrelated** or independent.
- Decorrelation of the tree means there is no connection between any two base-learners.
- Each base-learner is independently constructed and may have high variance.
- When many such Decorrelated trees are brought together, the overall variance tends to reduce.

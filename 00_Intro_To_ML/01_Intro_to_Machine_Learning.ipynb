#### Machine Learning

```
         Training Phase        │    Inference Phase
         --------------        │    ---------------
┌──────┐  ┌───────────┐        │       ┌───────┐
│ Data ├─>┤ Algorithm │        │       │ Input │
└──────┘  └─────┬─────┘        │       └───┬───┘
                ↓              │           ↓
        ┌───────┴────────┐     │     ┌─────┴──────┐
        │ Learns Pattern │     │     │  Model or  │
        └───────┬────────┘     │     │ Hypothesis │
                ↓              │     └─────┬──────┘
          ┌─────┴──────┐       │           ↓
          │  Model or  │       │       ┌───┴────┐
          │ Hypothesis │       │       │ Output │
          └────────────┘       │       └────────┘
```

- Ability to lean without being explicitly programmed.
- Machine Learning approaches a problem in two phases:
  1. Create or gather training data.
  2. Train a model based on some Algorithm using training dataset in **Training phase**.
  3. Predict Target value as output by giving Features as input to the model in **Inference phase**.

#### Learning Phase

Training the Model or Hypothesis is done in Learning or Training Phase.

#### Inference Phase

Final model will be used for inference in real time decision making.

## 2 Machine Learning

### 2.1 Introduction to ML

#### Tom Mitchel's Definition

Tom Mitchell defines machine learning as:

A computer program is said to learn from experience $E$ for a specific task $T$ and performance measure $P$, if its performance at $T$, as measured by $P$, improves with $E$.

#### Properties

- It is important to identify, define and distinguish below three components in an ML problem:
  - Task
  - Experience
  - Performance
- For a given Task (Business problem), learn from Experience (historical data) to improve Performance (quality of predictions).
- When we choose an algorithm, we want to improve the performance, for a given specific task with some experience.

#### Example

- **Task**: Decide if an Email is spam or ham.
- **Experience**: Old emails
- **Performance**: How effectively we can identify a spam email.

### 2.2 Types of ML Problems

#### Types of Machine Learning Algorithms

1. **Supervised Learning**: Learning from **Labeled dataset**.
   - Classification: Predicting discrete values from a collection of **class labels**.
   - Regression: Predicting continuous values within some range found in the **Target** variable of dataset.
   - Forecasting
2. **Unsupervised Learning**: Learning from **Unlabeled dataset**.
   - Recommender system
   - Clustering: Separating the data-points into **groups** without any explicit labels.
3. **Reinforcement Learning**: **Reward** based learning.

#### What is Regression?

##### Real value Prediction

$$
\large
D\,:=\,\biggl\{\bigl(x_i,\,y_i \bigr)\;;\;x_i\in\mathbb{R}^d\;;\;y_i\in\mathbb{R}^1\biggr\}_{i=1}^n
$$

#### What is Classification?

##### Types of Classification problems

1. Binary classification
2. Multi-class classification

##### Binary Classification

$$
\large
D\,:=\,\biggl\{\bigl(x_i,\,y_i \bigr)\;;\;x_i\in\mathbb{R}^d\;;\;y_i\in\bigl\{0,\,1\bigr\}\biggr\}_{i=1}^n
$$

##### Multi-class Classification

$$
\large
D\,:=\,\biggl\{\bigl(x_i,\,y_i \bigr)\;;\;x_i\in\mathbb{R}^d\;;\;y_i\in\bigl\{1,\,2,\,3\bigr\}\biggr\}_{i=1}^n
$$

> **Note**:
>
> In Supervised Machine Learning, Target will always be of one dimension i.e., single Target variable.

#### What is Forecasting?

Forecasting is the process of using historical and/or current information to estimate the value of a variable at a future point in time.

#### How Forecasting is different from Regression?

Forecasting predicts future values over time by focusing on sequential trends and historical patterns, whereas regression estimates relationships between a target variable and independent causal features regardless of time order

Types of Machine Learning Algorithms:

1. **Supervised Learning**: Learning from **Labeled dataset**.
   - Classification: Predicting discrete values from a collection of **class labels**.
   - Regression: Predicting continuous values within some range found in the **Target** variable of dataset.
   - Forecasting
2. **Unsupervised Learning**: Learning from **Unlabeled dataset**.
   - Recommender system
   - Clustering: Separating the data-points into **groups** without any explicit labels.
3. **Reinforcement Learning**: **Reward** based learning.

Types of Classification problems:

1. Binary classification
2. Multi-class classification

#### Regression

$$
\large
D\,:=\,\biggl\{\bigl(x_i,\,y_i \bigr)\;;\;x_i\in\mathbb{R}^d\;;\;y_i\in\mathbb{R}^1\biggr\}_{i=1}^n
$$

#### Classification

##### Binary Classification

$$
\large
D\,:=\,\biggl\{\bigl(x_i,\,y_i \bigr)\;;\;x_i\in\mathbb{R}^d\;;\;y_i\in\bigl\{0,\,1\bigr\}\biggr\}_{i=1}^n
$$

##### Multi-class Classification

$$
\large
D\,:=\,\biggl\{\bigl(x_i,\,y_i \bigr)\;;\;x_i\in\mathbb{R}^d\;;\;y_i\in\bigl\{1,\,2,\,3\bigr\}\biggr\}_{i=1}^n
$$

> **Note**:
>
> In Supervised Machine Learning, Target variable will always be of one dimension i.e., single value.

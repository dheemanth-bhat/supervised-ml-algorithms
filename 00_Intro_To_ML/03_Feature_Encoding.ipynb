## 1 Introduction

### 1.1 Introduction

#### What is Feature Encoding?

Feature encoding is the data preprocessing process of converting categorical or text-based data into numerical values so that machine learning algorithms can understand and process them.

#### Need for Feature Encoding 

- Categorical variables can be in non numerical values.
- ML Algorithms cannot understand non numerical data hence we need some techniques to convert them to numerical values.
- Feature Encoding **converts non-numerical data into numerical data**.

#### Types of Encoding

Some of the most common feature encoding techniques:

1. Label Encoding
2. One-Hot Encoding
3. Target Encoding
4. Ordinal Encoding

## 2 Label Encoding

### 2.1 Introduction

#### Properties

1. Label Encoding creates 1 column for each categorical features.
2. Each value in a category gets one numerical value.
3. The original feature is dropped after Label encoding.

#### Advantages

1. Simplest of all encoding techniques, simply replacing text with some number.
2. Original column is replaced with only one column thus not causing Curse of Dimensionality.

#### Disadvantages

1. Creates a **false ordinal relationship** between nominal categories i.e., false ranking assumptions.
2. Label Encoding is **logic less** i.e., the number assigned to a label has no logic/reasoning behind it.

#### API

````python
from sklearn.preprocessing import LabelEncoder

lbl_encoder = LabelEncoder()
````

### 2.2 Examples

#### Example #1

````python
X = ["A", "C", "A", "A", "C", "D", "D", "D", "B", "D"]
lbl_encoder.fit_transform(X)
````

## 3 One-Hot Encoding

### 3.1 Introduction

#### Properties

1. **Cardinality** is defined as total number of unique values in the feature.
2. One-Hot encoding creates $n$ columns for $n$ unique values in the feature.
3. One-Hot Encoding is preferred when a categorical feature has **low cardinality** (typically fewer than 10 unique values).
4. The **original feature is dropped** after one-hot encoding.
5. The new features created from one-hot encoding is called as **Dummy variables**.

#### Advantages

Feature Encoding **does not introduce false ordinal relationship** between nominal categories.

#### Disadvantages

High cardinality adds too many dimensions (columns) into dataset causing **Curse of Dimensionality**.

#### API

````python
from sklearn.preprocessing import OneHotEncoder

oht_encoder = OneHotEncoder()
````

### 3.2 Examples

#### Example #1

````python
X = np.array(["A", "C", "A", "A", "C", "E", "D", "D", "B", "D"]).reshape(-1, 1)
oht_encoder.fit_transform(X).toarray()
````

## 4 Target Encoding

### 4.1 Introduction

#### Properties

1. As the name suggests "Target" encoding, a categorical feature is encoded using Target variable based some logic.
2. The logic can be some sort of aggregation logic applied using Target variable for each category's group.

#### Advantages

1. Original column is **replaced with only one column** thus not causing Curse of Dimensionality.
2. Feature Encoding **does not introduce false ordinal relationship** between nominal categories.

#### Disadvantages

Requires **Controlled Target Encoding** otherwise can create data leakage between independent and dependent variables.

#### How to avoid data leakage?

```
      TRAINING DATA

        ↓       ↓
     Target Encoding
        │
        ↓
    Encoded X
        │
        ↓
      Model
```

#### API

````python
from sklearn.preprocessing import TargetEncoder

tgt_encoder = TargetEncoder(target_type="continuous")
````

### 4.2 Out-of-Fold Target Encoding

### 4.3 Examples

#### Example #1

````python
X = np.array(["A", "C", "A", "A", "C", "D", "D", "D", "B", "D"]).reshape(-1, 1)
y = np.array([100, 545, 164, 121, 599, 209, 236, 285, 777, 208])
tgt_encoder.fit_transform(X, y)
````

## 5 Ordinal Encoding

### 5.1 Introduction

#### Properties

1. Ordinal Encoding creates 1 column for each categorical features.
2. Each value in a category gets one numerical value.
3. The original feature is dropped after Ordinal encoding.

#### Advantages

1. Users can retain **true or custom ordinal relationship** between nominal categories.
2. Original column is replaced with only one column thus not causing Curse of Dimensionality.

#### API

````python
from sklearn.preprocessing import OrdinalEncoder

ord_encoder = OrdinalEncoder()
````

### 5.2 Examples

#### Example #1

````python
ord_encoder = OrdinalEncoder()
X = np.array(["A", "C", "A", "A", "C", "D", "D", "D", "B", "D"]).reshape(-1, 1)
ord_encoder.fit_transform(X)
````

#### Example #2

````python
# Retaining true/custom rank: ["D", "C", "B", "A"]
ord_encoder = OrdinalEncoder(categories=[["D", "C", "B", "A"]])
X = np.array(["A", "C", "A", "A", "C", "D", "D", "D", "B", "D"]).reshape(-1, 1)
ord_encoder.fit_transform(X)
````

### 5.3 LabelEncoder vs OrdinalEncoder

Difference between LabelEncoder and OrdinalEncoder:

- Ordinal encoding and label encoding both convert categorical text values into integers, but they differ primarily in their intended data types and Scikit-learn API usage.
- `LabelEncoder` is designed to encode one-dimensional target variables (y).
- `OrdinalEncoder` is designed to encode two-dimensional input features (X) and can handle multiple columns at once.

#### LabelEncoder

````python
lbl_encoder = LabelEncoder()
y = ["Yes", "No", "Yes", "Yes", "Yes", "No", "No"]
lbl_encoder.fit_transform(y)
````

#### OrdinalEncoder

````python
X = np.column_stack(
    [
        ["A", "C", "A", "A", "C", "B", "D"],
        ["Blr", "Hyd", "Blr", "Blr", "Chi", "Hyd", "Chi"],
    ]
)
X
````

````python
ord_encoder = OrdinalEncoder()
ord_encoder.fit_transform(X)
````

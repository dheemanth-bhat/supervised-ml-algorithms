### 1 Setup

#### Import and configure Libraries

```python
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Configure NumPy.
rng = np.random.default_rng(seed=42)
np.set_printoptions(linewidth=100)

# Configure Seaborn.
sns.set_style("whitegrid")
sns.set_palette("deep")
```

### 2 Heterogeneous Neighbors

#### Generate Features

```python
# X = rng.random(size=(250, 2))
X = rng.normal(loc=0.0, scale=3.5, size=(250, 2))
rng.shuffle(X)

display(X[:5])
X.shape
```

#### Generate Class Labels

```python
y = rng.integers(low=0, high=2, size=250)
rng.shuffle(y)

display(y[:5])
y.shape
```

#### Plot Heterogeneous Neighbors

```python
plt.figure(figsize=(6, 4))

sns.scatterplot(x=X[:, 1], y=X[:, 0], hue=y)
plt.title("Heterogeneous Neighbors")
plt.xlabel("X1")
plt.ylabel("X2")

plt.tight_layout()
# plt.savefig("images/heterogeneous_neighbors.png")
plt.show()
```

### 3 Outliers

#### Generate Features

```python
X = rng.normal(loc=0.0, scale=3.5, size=(50, 2))
rng.shuffle(X)

display(X[:5])
X.shape
```

#### Generate Class Labels

```python
mask = np.where((X < -3) | (X > 3), 0, 1)
y = mask[:, 0] * mask[:, 1]

display(y[:5])
y.shape
```

#### Plot Outlier Point Example #1

```python
sns.scatterplot(x=X[:, 1], y=X[:, 0], hue=y)
plt.title("Outlier of 'Class 0' inside 'Class 1'")
plt.plot(1.0, -1.0, color="blue", marker="o", markersize=6, label="Outlier of 'Class 0'")
plt.plot(1.3, -1.5, color="black", marker="o", markersize=6, label="Query Point")
plt.xlabel("X1")
plt.ylabel("X2")

plt.legend(loc="upper left")
plt.tight_layout()

# plt.savefig("images/outlier_in_neighborhood.png")
plt.show()
```

#### Generate Clusters

```python
# Generate clusters.
group1 = rng.normal(loc=2.0, scale=1.0, size=(30, 2))
group2 = rng.normal(loc=7.0, scale=1.0, size=(30, 2))
```

#### Plot Outlier Point Example #2

```python
plt.scatter(x=group1[:, 0], y=group1[:, 1], label="Class A")
plt.scatter(x=group2[:, 0], y=group2[:, 1], label="Class B")

plt.plot(3.7, 4.3, color="black", marker="o", markersize=7, label="Query Point")
plt.plot(4.0, 4.0, color="orange", marker="o", markersize=7, label="Outlier of 'Class B'")

plt.title("Outlier of a 'Class B' causing high variance")
plt.xlabel("X1")
plt.ylabel("X1")
plt.legend(loc="upper left")
plt.tight_layout()

# plt.savefig("images/outlier_high_variance.png")
plt.show()
```

### Underfitting

#### Generate Clusters

```python
# Generate clusters.

cls_a = rng.normal(loc=[0.0, 3.0], scale=0.5, size=(18, 2))
cls_b = rng.normal(loc=[0.0, 0.0], scale=1.0, size=(100, 2))
cls_c = rng.normal(loc=[3.0, 0.0], scale=0.5, size=(20, 2))
```

#### Plot Outlier Point Example #2

```python
plt.scatter(x=cls_a[:, 0], y=cls_a[:, 1], label="Class A")
plt.scatter(x=cls_b[:, 0], y=cls_b[:, 1], label="Class B")
plt.scatter(x=cls_c[:, 0], y=cls_c[:, 1], label="Class C")

plt.title("Large 'K' value causes high bias towards 'Class B'")
plt.xlabel("X1")
plt.ylabel("X1")
plt.legend(loc="upper right")
plt.tight_layout()

# plt.savefig("images/large_k_high_bias.png")
plt.show()
```

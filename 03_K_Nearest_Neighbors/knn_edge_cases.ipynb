### 1 Setup

#### Import and configure Libraries

```python
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Configure NumPy.
rng = np.random.default_rng(seed=42)
np.set_printoptions(linewidth=100)

# Configure Seaborn.
sns.set_style("whitegrid")
sns.set_palette("deep")
```

### 2 Heterogeneous Neighbors

#### Generate Features

```python
# X = rng.random(size=(250, 2))
X = rng.normal(loc=0.0, scale=3.5, size=(250, 2))
rng.shuffle(X)

display(X[:5])
X.shape
```

#### Generate Class Labels

```python
y = rng.integers(low=0, high=2, size=250)
rng.shuffle(y)

display(y[:5])
y.shape
```

#### Plot Heterogeneous Neighbors

```python
plt.figure(figsize=(6, 4))

sns.scatterplot(x=X[:, 1], y=X[:, 0], hue=y)
plt.title("Heterogeneous Neighbors")
plt.xlabel("X1")
plt.ylabel("X2")

plt.tight_layout()
# plt.savefig("images/heterogeneous_neighbors.png")
plt.show()
```

### 3 Outliers

#### Generate Features

```python
X = rng.normal(loc=0.0, scale=3.5, size=(50, 2))
rng.shuffle(X)

display(X[:5])
X.shape
```

#### Generate Class Labels

```python
mask = np.where((X < -3) | (X > 3), 0, 1)
y = mask[:, 0] * mask[:, 1]

display(y[:5])
y.shape
```

#### Plot Outlier Point

```python
plt.figure(figsize=(6, 4))

sns.scatterplot(x=X[:, 1], y=X[:, 0], hue=y)
plt.title("Outlier point in class 1")
plt.plot(1.0, -1.0, "bo", markersize=4, label="Outlier Point")
plt.plot(1.3, -1.5, "ko", markersize=4, label="Query Point")
plt.xlabel("X1")
plt.ylabel("X2")

plt.legend(loc="upper left")
plt.tight_layout()

# plt.savefig("images/outlier_in_neighborhood.png")
plt.show()
```

#### Import Libraries

```python
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

#### Configure Libraries

```python
# Configure NumPy.
rng = np.random.default_rng(seed=42)
np.set_printoptions(linewidth=100)

# Configure Seaborn.
sns.set_style("whitegrid")
sns.set_palette("deep")
```

#### Cluster #1

```python
# Number of points per cluster
n_points = 50

# Cluster 1 centered at (x=2, y=2)
cluster_1 = np.random.normal(loc=[6, 0], scale=1.0, size=(n_points, 2))
cluster_1[:5]
```

```python
# sns.histplot(x=cluster_1[:, 0])
# sns.histplot(x=cluster_1[:, 1])
# plt.show()
```

#### Cluster #2

```python
# Number of points per cluster
n_points = 50

# Cluster 2 centered at (x=6, y=6)
cluster_2 = np.random.normal(loc=[0, 6], scale=1.5, size=(n_points, 2))
cluster_2[:5]
```

```python
# sns.histplot(x=cluster_2[:, 0])
# sns.histplot(x=cluster_2[:, 1])
# plt.show()
```

#### Class Labels

```python
# Create labels for the points (0 for cluster1, 1 for cluster2)
labels = np.array([0] * n_points + [1] * n_points)
```

#### Plot Clusters

```python
# Combine the clusters into a single dataset
data = np.vstack((cluster_1, cluster_2))
```

```python
# Plot the points
sns.scatterplot(x=data[:, 0], y=data[:, 1], hue=labels)
plt.title("Two Clusters of Points")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

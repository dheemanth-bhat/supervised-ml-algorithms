# Training and Bias-Variance Tradeoff

## 1 Training

### 1.3 Data Sensitivity

#### Impact of High Dimensionality

- KNN may not work effectively when dataset has very high dimensions.
- Due to **Distance Concentration Effect** avoid using KNN on dataset with large number of features.
- For certain high-dimensional problems, Manhattan distance can preserve relative contrast better than Euclidean distance.

#### Impact of Outliers

- KNN is impacted by outliers.

#### Impact of Varied Scales

- KNN is a distance based algorithm hence it is impacted when feature are in different scales.
- Feature Scaling must be done before training a KNN model on Training dataset.

#### Need for Feature Encoding

- Feature Encoding must be done before training a KNN model on Training dataset.
- Avoid One-Hot encoding of features as it increases the number of dimensions in the dataset.

#### Impact of Imbalanced Dataset

- KNN is impacted by Imbalanced dataset.

## 2 Bias Variance Tradeoff

### 2.2 Underfitting

#### How to detect underfitting?

When the Training accuracy and Testing accuracy are very low.

#### Scenarios for underfitting

![Large K high Bias](images/large_k_high_bias.png)

Using a very large $K$ value for a severely imbalanced dataset causes underfitting. For example:

- If we have data-points such that: `{"Class A": 18, "Class B": 100, "Class C": 20}` and $K = 70$
- When $K = 70$, even if distance calculation covers all 18 points of "Class A", 20 points of "Class C",  
  "Class B" will still have majority of 32 votes creating an impact on final predictions.
- This results in every Query-point being predicted as "Class B".

### 2.3 Overfitting

#### How to detect underfitting?

When the Training accuracy is high but the Testing accuracy is relatively very low.

#### Scenarios for Overfitting

![Outlier causing high variance](images/outlier_high_variance.png)

Using a very small value of $K$ causes overfitting.

### 2.4 Hyperparameters

KNN has only one Hyperparameter $K$.

1. Very large $K$ value causes underfitting.
2. Very small $K$ value causes overfitting.

#### Summary

<div style="display: inline-block">

| Hyperparameter | Bias-Variance Tradeoff                     |
| :------------- | :----------------------------------------- |
| Large $K$      | High Bias, Low variance hence Underfitting |
| $K = 1$        | Low Bias, High Variance hence Overfitting  |

</div>

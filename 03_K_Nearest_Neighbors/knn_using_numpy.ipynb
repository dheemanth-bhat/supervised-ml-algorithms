```python
import numpy as np
```

```python
# Configure NumPy.
rng = np.random.default_rng(seed=42)
np.set_printoptions(linewidth=100)
```

### 1 Sample data

```python
data = np.array(
    [
        [3, 6, 1],
        [6, 4, 1],
        [8, 2, 3],
        [7, 5, 3],
        [1, 4, 2],
        [2, 2, 2],
    ]
)
data
```

#### Features

```python
X = data[:, 0:2]
X
```

#### Class Labels

```python
y = data[:, -1].reshape(-1, 1)
y
```

#### Query Point

```python
qp = [2, 5]
```

### 2 KNN Algorithm

#### STEP #1: Distance b/w query-point & data-points

```python
distances = np.linalg.norm(X - qp, ord=2, axis=1)
distances.tolist()
```

#### STEP #2: Sort distances in ascending order

```python
mask = distances.argsort(descending=False)
nearest_neignbors = y[mask]

print("Class labels of nearest to farthest neighbors:")
nearest_neignbors
```

#### STEP #3: Pick K number of Neighbors

```python
k = 3
k_nearest_neignbors = nearest_neignbors[:k]
k_nearest_neignbors
```

#### STEP #4: Vote to identify majority class-label

```python
def mode(x):
    values, counts = np.unique(x, return_counts=True)
    mask = np.argmax(counts)
    majority_cl = values[mask]

    return majority_cl
```

```python
y_pred = mode(k_nearest_neignbors)
print("Predicted class label:", y_pred.item())
```

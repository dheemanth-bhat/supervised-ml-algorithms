```python
import numpy as np
```

```python
# Configure NumPy.
rng = np.random.default_rng(seed=42)
np.set_printoptions(linewidth=100)
```

#### Sample Data-points

```python
data = np.array(
    [
        [3, 6, 1],
        [6, 4, 1],
        [8, 2, 3],
        [7, 5, 3],
        [1, 4, 2],
        [2, 2, 1],
    ]
)
data
```

```python
x = data[:, 0:2]
x
```

```python
y = data[:, -1]
y
```

```python
qp = [2, 5]
```

#### Calculate Euclidean Distance

```python
dist = np.linalg.norm(x - qp, ord=2, axis=1)
dist
```

#### Concat Distances and Class-labels

```python
def concate_cols(c1, c2):
    c1 = c1.reshape(-1, 1)
    c2 = c2.reshape(-1, 1)

    return np.hstack((c1, c2))
```

```python
neignbors = concate_cols(dist, y)
neignbors
```

#### Sort Distances

```python
neignbors[:, 0].argsort()
```

```python
def sort_by(x, idx: int, descending=False):
    return x[x[:, idx].argsort(descending=descending)]
```

```python
sorted_neignbors = sort_by(neignbors, 0)
sorted_neignbors
```

#### Vote for Class-labels

```python
def value_count(x, idx):
    labels, votes = np.unique(x[:, idx], return_counts=True)
    return concate_cols(labels, votes)
```

```python
k = 3
votes = value_count(sorted_neignbors[:k], 1)
top_k = sort_by(votes, 1, descending=True)
top_k
```

```python
y_pred = top_k[0][0]
print("Predicted class label:", y_pred.item())
```

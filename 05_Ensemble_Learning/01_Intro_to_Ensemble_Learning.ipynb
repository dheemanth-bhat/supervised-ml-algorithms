# Intro to Ensemble Learning Techniques

## 1 Ensemble Learning

Ensembling short for Ensemble Learning.

### 1.1 Need for Ensembling

#### Problem Statement

- Decision Tree models are prone to overfitting.
- We need a Tree that can generalize well on unseen data.
- Pruning, cross-validation and Hyperparameter tuning on a single model can reduce overfitting to some extent.
- There is a need for more robust techniques that can reduce the impact of overfitting and improve prediction results.

#### Solution

- Instead of finalizing a prediction from just one model, train multiple models and combine their predictions.
- Combine predictions from multiple models to arrive at final prediction.
- **Ensembling** is a technique that combines multiple individual models into a single, more accurate and reliable predictive model.

### 1.2 Components of Ensembling

#### 1 Base Learners

- Individual machine learning models such as decision trees, linear regression, or support vector machines that make initial predictions.
- Base Learners are allowed to be overfitting or underfitting.
- Base Learners are also called as Weak learners.
- We can have any number of Base Learners and are decided based on Hyperparameter.

#### 2 Diversity Mechanism

- A strategy to ensure base learners make different types of errors.
- Diversity can be achieved by splitting the training data randomly or by changing how base-learners learn.

#### 3 Aggregation

- The rule or secondary model used to merge outputs from all base learners into one final decision.
- Types of Aggregation techniques:
  1. Majority Voting
  2. Averaging
  3. Meta-Learner

### 1.3 Types of Ensembling Techniques

#### Based on Base learner swarm

1. Heterogeneous Ensembling
2. Homogeneous Ensembling

#### Based on Ensembling strategy

1. Bagging
2. Boosting
3. Stacking

### 1.4 Heterogeneous Ensembling

#### Example #1

Smaller samples from dataset are used to train base learners on different ML algorithm.

```
          Entire Training Dataset
                     │
      ┌──────────────┼────────────────┐
      ↓              ↓                ↓
  Sample #1      Sample #2        Sample #3
      ↓              ↓                ↓
Decision Tree       KNN       Logistic Regression
      │              │                │
      └──────────────┼────────────────┘
                     ↓
                Aggregation
                     ↓
             Final Prediction
```

## 2 Bagging

### 2.1 Introduction

Full form of Bagging is Bootstrapped Aggregation:

1. Bootstrapping: Random sampling (with replacement) of rows and columns from a population dataset for individual models.
2. Aggregation: Aggregating the prediction results of different models.

#### Steps in Bagging

1. **Multiple random samples** are drawn from the same dataset.
1. These samples have **random features** selected from the same dataset.
1. A swarm of models called **Base Learners** are **trained in parallel**.
1. Each base learner is trained on set of samples (Features and Target).
1. Each base learner will produce one prediction as output.
1. Finally take the average of all the output predictions from the swarm of models.

#### Properties

- In Bagging, models are trained in parallel independently of one another.
- Models can be trained on different ML algorithms and different samples from same dataset.
- Multiple models can be trained using same ML algorithm but on different samples from same dataset.
- Bagging is like performing K-Fold CV but on different models.
- Aggregation causes canceling of errors thus minimizing overall error.

### 2.2 Random Forest

#### Ensembling

```
            Entire Training Dataset
                       │
      ┌────────────────┼────────────────┐
      ↓                ↓                ↓
  Sample #1        Sample #2        Sample #3
      ↓                ↓                ↓
Decision Tree    Decision Tree    Decision Tree
      │                │                │
      └────────────────┼────────────────┘
                       ↓
                  Aggregation
                       ↓
               Final Prediction
```

#### Components of Ensembling in RF

##### 1 Base Learner

Random Forest is a special type of Bagging technique where every base learner in the swarm is a Decision Tree.

##### 2 Diversity Mechanism

- Diversity is achieved using Bootstrapping.
- Bootstrapping is sampling with replacement of different subset of features.

##### 3 Aggregation

- For Classification Tasks - The model uses majority voting (or the mode).
- For Regression Tasks - The model uses averaging (or mean).

#### Properties

- Random Forest is a type of Homogeneous Ensembling where every base learner is a Decision Tree.
- Random Forest is the most popular and proven effective type of Bagging.

### 2.3 Issues with Bagging

#### Disadvantages

- Bagging is not very effective when we have less data.
- Bagging is not very effective when dataset has complex patterns.

## 3 Boosting

### 3.1 Introduction

Boosting is an ensemble learning technique that sequentially combines a set of weak, low-accuracy models (called weak learners) to create a single, highly accurate model (a strong learner).

#### Idea of Boosting

- Start with a simple (high bias, low variance a.k.a underfit) model. An underfit model will have high Training error.
- Build sequence of models with the goal to slightly reduce the error of previous model. Newer models are expected to have lesser bias.
- Boost the prediction of previous model by reducing the errors in next model.
- Hence the goal of Boosting is to reach an optimal prediction for every model.

#### Properties

- In Boosting, models are trained in sequence using the errors of previous model.
- Boosting is also known as Additive Combination.
- Boosting can learn the non-linear patterns in the data.

> **Note**:
> 
> Best example for a high bias, low variance model is - Mean model (which predict mean value every time).

### 3.2 Components of Boosting

#### High Bias Model

#### Error Component

$$
\large
y\;=\;\hat{y}\;+\;\text{error}
$$

- Ideally actual Target value $y$ should be equal to predicted Target value $\hat{y}$ plus some error in prediction.
- This predicted Target value $\hat{y}$ comes from high bias model.
- This $\text{error}$ component is predicted by a separate model.
- End goal of Boosting is to reach an optimal prediction for every model.

### 3.3 Steps in Boosting

#### STEP #1: Simple Model

- Start with a simple (high bias, low variance a.k.a underfit) model.
- Predict target values using this high bias model.
- Prediction at this stage zero is represented as:

$$
\large
y_0\;=\;h_0(x)
$$

#### STEP #2: Error Prediction Model

- Build a new model that **predicts the errors** - $h(x)$ - of previous model.
- Choose an appropriate Learning Rate - $\gamma$ - to control the impact of predicted errors.
- Add the predicted errors (controlled using Learning Rate) to the predictions from previous model.

$$
\large
h_{i - 1}(x)\;+\;\gamma \cdot h_i(x)
$$

#### STEP #3: Sequential Training

Repeat STEP #2 - $m$ - number of times to have sequence of models until a final model has low bias.

$$
\large
\begin{aligned}
F_m(x)\;&=\;\Biggl(\biggl(\Bigl(\bigl(h_0(x)\;+\;\gamma_1\,h_1(x)\bigr)\;+\;\gamma_2\,h_2(x)\Bigr)\;+\;\gamma_3\,h_3(x)\biggr)+\ldots\Biggr)\;+\;\gamma_m\,h_m(x) \\[10pt]
        &\qquad\text{Simplify above equation using associative property of addition} \\[10pt]
        &=\;h_0(x)\;+\;\gamma_1\,h_1(x)\;+\;\gamma_2\,h_2(x)\;+\;\gamma_3\,h_3(x)\;+\;\ldots\;+\;\gamma_m\,h_m(x) \\[10pt]
F_m(x)\;&=\;h_0(x)\;+\;\sum_{i\,=\,1}^m\,\gamma_i\,h_i(x)
\end{aligned}
$$

### 3.4 Advantages and Disadvantages

#### Advantages

- Boosting is useful when we have complex data with complex patterns.
- Boosting is very effective when we have less data.

#### Disadvantages

- Boosting is slower than Bagging because of the sequential modeling.

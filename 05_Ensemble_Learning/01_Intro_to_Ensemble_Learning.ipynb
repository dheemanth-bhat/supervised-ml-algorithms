# Intro to Ensemble Learning Techniques

## 1 Ensemble Learning

### 1.1 Need for Ensemble Learning

#### Problem Statement

- Decision Tree models are prone to overfitting.
- We need a Tree that can generalize well on unseen data.
- Pruning, cross-validation and Hyperparameter tuning on a model can reduce overfitting to some extent.
- There is a need for more robust techniques that can reduce the impact of overfitting on prediction results.

#### Solution

- Instead of finalizing a prediction from just one model, train multiple models and combine their predictions.
- Combine predictions from multiple models to arrive at final prediction.
- **Ensembling** is a technique that combines multiple individual models into a single, more accurate and reliable predictive model.

#### Types of Ensembling Techniques

1. Bagging
2. Boosting
3. Stacking

### 1.2 Bagging

Full form of Bagging is Bootstrapped Aggregation:

1. Bootstrapping: Random sampling (with replacement) of rows and columns from a population dataset for individual models.
2. Aggregation: Aggregating the prediction results of different models.

#### Steps in Bagging

1. **Multiple random samples** are drawn from the same dataset.
1. These samples have **random features** selected from the same dataset.
1. A swarm of models called **Base Learners** are **trained in parallel**.
1. Each base learner is trained on set of samples (Features and Target).
1. Each base learner will produce one prediction as output.
1. Finally take the average of all the output predictions from the swarm of models.

#### Properties

- Models can be trained on different ML algorithms and different samples from same dataset.
- Multiple models can be trained using same ML algorithm but on different samples from same dataset.
- Bagging is like performing K-Fold CV but on different models.
- Aggregation causes canceling of errors thus minimizing overall error.

#### Types of Bagging Techniques

1. Heterogeneous Ensembling
2. Homogeneous Ensembling

### 1.3 Heterogeneous Ensembling

#### Example #1

Bootstrapped samples are used to train base learners on different ML algorithm

```
          Entire Training Dataset
                     │
      ┌──────────────┼────────────────┐
      ↓              ↓                ↓
  Sample #1      Sample #2        Sample #3
      ↓              ↓                ↓
Decision Tree       KNN       Logistic Regression
      │              │                │
      └──────────────┼────────────────┘
                     ↓
                Aggregation
                     ↓
             Final Prediction
```

#### Example #2

Same dataset (entire training dataset) is used to train base learners trained on different ML algorithm.

```
          Entire Training Dataset
                     │
      ┌──────────────┼──────────────┐
      ↓              ↓              ↓
Decision Tree       KNN            SVM
      │              │              │
      └──────────────┼──────────────┘
                     ↓
                Aggregation
                     ↓
             Final Prediction
```

### 1.4 Random Forest

#### Ensembling

```
            Entire Training Dataset
                       │
      ┌────────────────┼────────────────┐
      ↓                ↓                ↓
  Sample #1        Sample #2        Sample #3
      ↓                ↓                ↓
Decision Tree    Decision Tree    Decision Tree
      │                │                │
      └────────────────┼────────────────┘
                       ↓
                  Aggregation
                       ↓
               Final Prediction
```

Random Forest is a special type of Bagging technique where every model in the swarm is a Decision Tree.

#### Properties

- Random Forest is a type of Homogeneous Ensembling where every base learner is a Decision Tree.
- Random Forest is the most popular and proven effective type of Bagging.

## 1 Derivative of Loss function

### 1.1 Loss Function in Linear Regression

#### Mean Squared Error

- Mean Squared Error is chosen as Loss function during optimization of Linear Regression model.
- MSE is an error metric and we have to minimize it hence the goal of optimization is to minimization the loss function.
- LR learns coefficients (weights) and an intercept by minimizing the difference between target's actual and predicted values.

#### Why not MAE or RMSE?

- Absolute operation in Mean Absolute Error (MAE) makes it non differentiable.
- Square root operation in Root Mean Squared Error (RMSE) makes it complex to differentiable.

#### Formula

$$
\large
\mathrm{MSE}\;=\;\frac{1}{n}\sum_{i\,=\,1}^{n}\;\Bigl(y^{(i)}\;-\;\hat{y}^{(i)}\Bigr)^2
$$

Where,

- $\mathrm{MSE}$ - Mean Squared Error
- $n$ - Total number of rows in the dataset $D$
- $y^{(i)}$ - Target's actual value for $i^\text{th}$ row in the dataset
- $\hat{y}^{(i)}$ - Target's predicted value for $i^\text{th}$ row in the dataset

### 1.2 Optimization of Loss function

#### STEP #1: Define Loss function

Loss function $l(D,\,\vec{w},\,w_0)$ is defined as:

$$
\large
l(D,\,\vec{w},\,w_0)\;=\;\frac{1}{n}\sum_{i\,=\,1}^{n}\;\Bigl(y^{(i)}\;-\;\hat{y}^{(i)}\Bigr)^2
$$

#### STEP #2: Define Minimization of Loss function

Goal of optimization is to minimize loss function: $l(D,\,\vec{w},\,w_0)$

$$
\large
\begin{aligned}
w^*,\,w_0^* &= \arg\min_{w,\,w_0}\;\;l(D,\,\vec{w},\,w_0) \\[5pt]
            &= \arg\min_{w,\,w_0}\;-\frac{1}{n}\,\sum_{i\,=\,1}^n\,\Bigl(y^{(i)}\;-\;\hat{y}^{(i)}\Bigr)^2
\end{aligned}
$$

Where,

- $w^*$ - Optimal weight vector
- $w_0^*$ - Optimal Bias

#### STEP #3: Calculate derivatives of Loss function

Partial derivatives of loss function $l$ w.r.t weight vector $w$:

$$
\newcommand{\pdl}[1]{\frac{\partial}{\partial\,{#1}} l}
\large
\begin{aligned}
\pdl{w}\;&=\; -2\,\Bigl(y^{(i)}\;-\;\hat{y}^{(i)}\Bigr) \cdot x^{(i)} \\[8pt]
\pdl{w_0}\;&=\; -2\,\Bigl(y^{(i)}\;-\;\hat{y}^{(i)}\Bigr)
\end{aligned}
$$

#### STEP #4: Apply Gradient Update Rule (GUR)

$$
\large
w^{\text{new}}\;=\;w^{\text{old}}\;-\;\eta\,\cdot\frac{\partial}{\partial\,w} l(D,\,\vec{w},\,w_0)
$$

Where,

- $w^{\text{new}}$ - Weight vector after update
- $w^{\text{old}}$ - Weight vector before update
- $\eta$ - Learning Rate

#### STEP #5: Gradient Update Rule for LR

$$
\newcommand{\residual}{\frac{1}{n}\sum_{i\,=\,1}^n\,-2\,\Bigl(y^{(i)}\;-\;\hat{y}^{(i)}\Bigr)}
\large
\begin{aligned}
  w^{\text{new}}\;&=\;w^{\text{old}}\;-\;\eta\,\Biggl[\residual \cdot x^{(i)}\Biggr] \\[10pt]
w_0^{\text{new}}\;&=\;w_0^{\text{old}}\;-\;\eta\,\Biggl[\residual\Biggr]
\end{aligned}
$$

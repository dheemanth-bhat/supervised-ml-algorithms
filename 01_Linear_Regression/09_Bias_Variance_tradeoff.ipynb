# Bias-Variance Trade off

## 1 Introduction

### 1.1 Generalization

#### Degradation

Degradation in ML is the decline in a model's accuracy, reliability, or performance over time after it has been deployed into a live production environment.

#### Degradation b/w Train and Test

- Degradation between train and test refers to the decrease in a machine learning model's performance when it is evaluated on unseen test data compared to its performance on the training data.
- Degradation indicates model is unable to reproduce what it has learnt from training data, instead it is degrading.

#### Generalization vs Degradation

- Generalization is said to be achieved if the Degradation between Train and Test accuracy is as close enough as possible.
- The goal is to choose the model that generalizes well (i.e., least degradation).
- When an ML model fails to generalize, it typically  results in Underfitting or Overfitting.

### 1.2 Underfitting vs Overfitting

#### Underfitting

- When both Train-error and Test-error are high model is said to be underfit on Training data.
- "Not enough" learning.

#### Overfitting

- When Train-error is low but Test-error is high model is said to be overfit on Training data.
- Over learning or memorizing the training data.

## 2 Bias-Variance Trade off

### 2.1 Introduction

#### Definition

The Bias-Variance trade-off is a fundamental dilemma in ML to balance two types of error that prevent an algorithm from generalizing well on unseen data.

#### Types of Errors

There are two different types of errors in a model's predictions:

1. Bias
2. Variance

#### Formula

$$
\large
\text{Total Error}\;=\;\text{Bias}^2\;+\;\text{Variance}\;+\;\text{Irreducible Error}
$$

### 2.2 Bias

#### Definition

**Bias** is defined as systematic **error** that occurs when a model makes overly simplistic assumptions about the data, causing it to miss important patterns and Underfit.

#### What does error in Bias definition mean?

- Error in Bias definition represents both Train-error and Test-error.
- High Bias mean high Train-error and high Test-error hence model is Underfitting.

#### Properties

- Bias: How far the model's average prediction is from the true underlying relationship.
- In other words Bias is the measure of **How close a model is to reality**.
- An ideal model as high accuracy or low error. Low error means low bias.
- **High Bias indicates Underfitting**.

### 2.3 Variance

#### Definition

Variance measures how much a model's predictions change when it is trained on different subsets or variations of the training data.

#### Properties

- Variance: How much the model's predictions change when it is trained on different training datasets.
- **High Variance indicates Overfitting**.

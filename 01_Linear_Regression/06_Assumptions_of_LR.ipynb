## 3 Test for No Multicollinearity

### 3.1 Introduction

#### Analytical Tests

1. Variance Inflation Factor (VIF)

### 3.2 Variance Inflation Factor

$$
\large
\mathrm{VIF}\;=\;\frac{1}{1\;-\;R_j^2}
$$

Where,

- $R_j^2$ is the Ordinary R-squared value of a particular, say $j^{\text{th}}$, independent variable and not the whole model.

#### API

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

#### Properties

- VIF falls in the range $[1,\,\infty)$

```python
for r2 in [-1_000, -100, -10, -1, -0.9, -0.5, -0.01, 0, 0.01, 0.5, 0.8, 0.9, 1]:
    vif = "infty" if r2 == 1 else round(1 / (1 - r2), 2)
    print(f"R-squared: {r2:>5}\t VIF: {vif:>5}")
```

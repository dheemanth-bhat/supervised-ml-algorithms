# Multi Variable Linear Regression

## 1 Advance concepts

### 1.1 Model Explainability

- The biggest advantage of Linear Regression is Model Explainability.
- Model coefficients tells us how important a feature is in predicted the value of $y$.

#### How to interpret weights?

For example if the linear relation between dependent variable $y$ and independent variables: $\{x_1,\,x_2,\,\ldots,\,x_n\}$ is:

$$
\large
y\;=\;w_1\,x_1\;+\;-w_2\,x_2\;+\;\ldots\;+\;w_n\,x_n\;+\;w_0
$$

Then weighs can be interpreted as:

- If $x_1$ increases by one unit then $y$ increases by $w_1$ units, assuming all other independent variables are constant.
- If $x_2$ increases by one unit then $y$ decreases by $w_2$ units, assuming all other independent variables are constant.

#### Feature Importance vs Coefficients

- Model coefficients can be used to determine feature importance.
- Higher the magnitude of coefficients (ignoring the sign), higher will be the importance of their corresponding feature in determining the target value.
- Sign tells the direction of correlation:
  - +ve sign indicates positive correlation i.e., $y$ increases as $x$ increases.
  - -ve sign indicates negative correlation i.e., $y$ decreases as $x$ increases.

#### Impact of Feature Scaling

- Feature scaling is done to give **equal importance to all features** in different scales.
- After Feature Scaling, raw coefficients no are longer directly comparable i.e., model interpretability is lost.
- Feature scaling can help **compare the magnitudes of different coefficients** by bringing them all to same scale.
- Feature scaling smoothens the Gradient Descent and is not over influenced by a feature in larger scale.

### 2.1 Optimization of Loss function

#### Compute Derivative of Loss function

#### Gradient Update Rule for LR

$$
\large
\begin{aligned}
w^{\text{new}}\;&=\;w^{\text{old}}\;-\;\eta\,\Biggl[\frac{1}{n}\sum_{i\,=\,1}^n\,-2\,\bigl(y_i\;-\;\hat{y}_i\bigr) \cdot x_i\Biggr] \\[10pt]
w_0^{\text{new}}\;&=\;w_0^{\text{old}}\;-\;\eta\,\Biggl[\frac{1}{n}\sum_{i\,=\,1}^n\,-2\,\bigl(y_i\;-\;\hat{y}_i\bigr)\Biggr]
\end{aligned}
$$

#### Simplified GUR for LR

$$
\large
\begin{aligned}
w^{\text{new}}\;&=\;w^{\text{old}}\;+\;\frac{2\eta}{n}\,\Biggl[\,\sum_{i\,=\,1}^n\,\bigl(y_i\;-\;\hat{y}_i\bigr) \cdot x_i\,\Biggr] \\[10pt]
w_0^{\text{new}}\;&=\;w_0^{\text{old}}\;+\;\frac{2\eta}{n}\,\Biggl[\,\sum_{i\,=\,1}^n\,\bigl(y_i\;-\;\hat{y}_i\bigr)\,\Biggr]
\end{aligned}
$$

```python
from sklearn.metrics import r2_score
```

```python
import numpy as np
def calc_r2(y_true, y_pred):
    
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
   
    y_mean = np.mean(y_true)

    #Calculate total sum of squares
    tot_ss = np.sum((y_true - y_mean) ** 2)
    
    #Calculate residual sum of squares
    res_ss = np.sum((y_true - y_pred) ** 2)
    
    #Calculate r2
    r2 = 1 - (res_ss / tot_ss)
    
    return r2.round(2)
```

```python
calc_r2([9.91, 12.79, 12.36, 5.31, 4.11],
[0.88, 2.6, 0.36, -0.45, 1.64])
```

```python
r2_score([9.91, 12.79, 12.36, 5.31, 4.11], [0.88, 2.6, 0.36, -0.45, 1.64])
```

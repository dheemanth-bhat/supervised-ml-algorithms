# Regularization

## 1 Introduction

### 1.1 Overfitting Problem

Overfitting is a major problem in Machine Learning that prevents model from Generalizing well on unseen data.

#### What causes Overfitting?

1. Over learning: Over optimized weight vectors.
2. Higher order Polynomial features.
3. Imbalanced dataset: Bias towards majority class.
4. Unshuffled and/or non-random Train-Test split.

#### How to resolve overfitting?

1. **Dropping unnecessary features** used in training.
2. **Data Augmentation**: Artificially expands the training set by creating modified copies of data.
3. **Reducing the degree** of Polynomial features.
4. Reducing the impact of a feature by **regulating the magnitude** of its coefficients/weights.

### 1.2 Regularization

#### Definition

Regularization is a set of techniques used in machine learning to prevent overfitting and improve a model's ability to generalize to new, unseen data.

#### How Regularization works?

- Regularization adds a penalty weights - $w$ - to the cost function to discourage overly complex models with large weights.
- Regularization cannot pin point a particular feature causing the overfitting.
- Regularization instead penalizes the weight vector to counterbalance the weights of overfitting features.

#### Example

For example given mapping function with polynomial features:

$$
\large
\hat{y}\;=\;f(x)\;=\;w_1\,x_1\;+\;w_2\,x_1^4\;+\;w_3\,x_1^2\;+\;w_4\,x_1^3\;+\;w_0
$$

if one the features, say $x_1^4$, is causing the model to overfit then its corresponding weights $w_2$ must be reduced.

#### Types of Regularization

There are three main types of Regularization techniques:

1. L1 Regularization (Lasso Regression)
2. L2 Regularization (Ridge Regression)
3. Elastic Net

### 1.3 Mathematical Intuition

#### Mapping Function

Given the Mapping function $f(x)$:

$$
\large
\hat{y}\;=\;f(x)\;=\;w_1\,x_1\;+\;w_2\,x_2\;+\;w_3\,x_3\;+\;w_4\,x_4\;+\;w_0
$$

Where,

- $\hat{y}$ - Predicted value
- $w$ - Weight vector
- $x$ - Feature vector

#### Old Cost Function (without Regularization)

Given the Cost function - $\mathrm{MSE}$ - Mean Squared Error:

$$
\large
\mathrm{MSE}\;=\;\frac{1}{n}\sum_{i\,=\,1}^n\,(y_i\;-\;\hat{y}_i)^2
$$

Where,

- $n$ - Total number of rows in the dataset (feature vectors)
- $y_i$ - Target's actual value for $i^\text{th}$ feature vector
- $\hat{y_i}$ - Target's predicted value for $i^\text{th}$ feature vector

#### New Cost Function (with Regularization)

Regularization adds penalty - $w$ - to Cost function to regulate the impact of unwanted weights:

$$
\large
\mathrm{MSE}\;=\;\frac{1}{n}\sum_{i\,=\,1}^n\,\bigl(\hat{y}_i\;-\;y_i\bigr)^2\;+\;w
$$

### 1.4 Constraint Optimization

#### Equation

Optimization of new cost function (with regularization):

$$
\large
w^*,\,w_0^*\;=\;\arg\min_{w,\,w_0}\,\Biggl[-\frac{1}{n}\sum_{i\,=\,1}^n\,\bigl(\hat{y}_i\;-\;y_i\bigr)^2\;+\;w\Biggr]
$$

Where,

- $w^*$ - Optimal weight vector
- $w_0^*$ - Optimal Bias

#### How Constraint Optimization works?

- Adding a constraint (in this case penalty) to the optimization of Cost function.
- Along with minimizing the original cost function the penalty must also me minimized.
- This constraint sort of creates a **circular layers** in the shape of original loss function.
- We can then choose the depth up to which the optimization can be allowed, hence controlling **over learning** or over optimization.
- Control on constraint is achieved using Lambda - $\lambda$ - called as **Regularization parameter**.

#### Lambda (Regularization parameter)

$$
\newcommand{\MSE}{\frac{1}{n}\sum_{i\,=\,1}^n\,\bigl(\hat{y}_i\;-\;y_i\bigr)^2}
\large
w^*,\,w_0^*\;=\;\arg\min_{w,\,w_0}\,\Biggl[-\MSE\;+\;\lambda\cdot{w}\Biggr]
$$

Where,

- $\lambda$ - Hyperparameter called as Regularization parameter to control Regularization

### 1.5 Model Complexity

#### How Lambda works?

When $\lambda$ value is:

- Zero no Regularization hence no penalization.
- Low (small value greater than zero), Regularization starts taking effect (low penalization).
- Very large, high Regularization leading to extreme penalization.

Typically try sequence of values, b/w $[0.01,\,1]$, while tuning for $\lambda$ hyperparameter. For example: $[0.01,\,0.05,\,0.1,\,0.5,\,1]$

#### Regularization vs Model Complexity

- No or Low Regularization -> Complex model -> Overfitting
- High Regularization -> Simple model -> Underfitting

#### Summary

<div style="display: inline-block">

| Lambda value        | Regularization | Complexity | Fitting      |
| :------------------ | :------------- | :--------- | :----------- |
| Nearly equal to 0   | Low            | High       | Overfitting  |
| Well tuned, say 0.1 | Moderate       | Moderate   | Best fit     |
| Large value, say 1  | High           | Low        | Underfitting |

</div>

## 2 Type of Regularization

### 2.1 Lasso Regression

#### Equation

$$
\newcommand{\MSE}{\frac{1}{n}\sum_{i\,=\,1}^n\,\bigl(\hat{y}_i\;-\;y_i\bigr)^2}
\newcommand{\Lasso}{\lambda\sum_{j\,=\,1}^d\,\vert{w_j}\vert}
\large
\text{Lasso Regression}\;=\;\arg\min_{w,\,w_0}\,\Biggl[-\MSE\;+\;\Lasso\Biggr]
$$

Where,

- $n$ - Total number of rows in the dataset (feature vectors)
- $d$ - Total number of columns in the dataset (dimension of a feature vector)
- $\mathrm{MSE}$ - Cost function (Mean Squared Error)
- $y_i$ - Target's actual value for $i^\text{th}$ feature vector
- $\hat{y_i}$ - Target's predicted value for $i^\text{th}$ feature vector
- $\lambda$ - Hyperparameter called as Regularization parameter to control effect of L1 Regularization

#### Properties

- Lasso Regression can force the weights to become zero.
- Hence Lasso Regression **can eliminate features** by shrinking their coefficients to exactly zero.

#### API

```python
from sklearn.linear_model import Lasso
```

### 2.2 Ridge Regression

#### Equation

$$
\newcommand{\MSE}{\frac{1}{n}\sum_{i\,=\,1}^n\,\bigl(\hat{y}_i\;-\;y_i\bigr)^2}
\newcommand{\Ridge}{\lambda\sum_{j\,=\,1}^d\,w_j^2}
\large
\text{Ridge Regression}\;=\;\arg\min_{w,\,w_0}\,\Biggl[-\MSE\;+\;\Ridge\Biggr]
$$

Where,

- $n$ - Total number of rows in the dataset (feature vectors)
- $d$ - Total number of columns in the dataset (dimension of a feature vector)
- $\mathrm{MSE}$ - Cost function (Mean Squared Error)
- $y_i$ - Target's actual value for $i^\text{th}$ feature vector
- $\hat{y_i}$ - Target's predicted value for $i^\text{th}$ feature vector
- $\lambda$ - Hyperparameter called as Regularization parameter to control effect of L2 Regularization

#### Properties

- Ridge Regression shrinks the coefficients close to zero, but **retains all features**.
- Faster compared to Lasso because Closed-form solution is available.

#### API

```python
from sklearn.linear_model import Ridge
```

### 2.3 Elastic Net

#### Equation

$$
\newcommand{\MSE}{\frac{1}{n}\sum_{i\,=\,1}^n\,\bigl(\hat{y}_i\;-\;y_i\bigr)^2}
\newcommand{\Lasso}{\lambda_1\sum_{j\,=\,1}^d\,\vert{w_j}\vert}
\newcommand{\Ridge}{\lambda_2\sum_{j\,=\,1}^d\,w_j^2}
\large
\text{Elastic Net}\;=\;\arg\min_{w,\,w_0}\,\Biggl[-\MSE\;+\;\Lasso\;+\;\Ridge\Biggr]
$$

Where,

- $n$ - Total number of rows in the dataset (feature vectors)
- $d$ - Total number of columns in the dataset (dimension of a feature vector)
- $\mathrm{MSE}$ - Cost function (Mean Squared Error)
- $y_i$ - Target's actual value for $i^\text{th}$ feature vector
- $\hat{y_i}$ - Target's predicted value for $i^\text{th}$ feature vector
- $\lambda_1$ - Hyperparameter to control effect of L1 Regularization
- $\lambda_2$ - Hyperparameter to control effect of L2 Regularization

#### Properties

- Elastic Net is a combination of Lasso Regression and Ridge Regression.

#### API

```python
from sklearn.linear_model import ElasticNet
```

### Formula #1

$$
\large
G(S)\;=\;1\,-\,\sum_{i\,=\;1}^c\,p_i^2
$$

#### Formula #1 Code

```python
def gini_impurity_f1(probs):
    """
    Function to calculate Gini Impurity using Formula #1.
    """
    tot = 0
    for p_i in probs:
        tot += p_i**2
    gi = round(1 - tot, 2)

    return gi
```

#### CASE #1: Sum of class probabilities is 1

```python
gini_impurity_f1([0.1, 0.2, 0.7])
```

#### CASE #2: Sum of class probabilities is not 1

```python
gini_impurity_f1([0.1, 0.2, 0.3])
```

### Formula #2

$$
\large
G(S)\;=\;\sum_{i\,=\;1}^c\,p_i\,(1 - p_i)
$$

#### Formula #2 Code

```python
def gini_impurity_f2(probs):
    """
    Function to calculate Gini Impurity using Formula #2.
    """
    gi = 0
    for p_i in probs:
        gi += p_i * (1 - p_i)

    return round(gi, 2)
```

#### CASE #1: Sum of class probabilities is 1

```python
gini_impurity_f2([0.1, 0.2, 0.7])
```

#### CASE #2: Sum of class probabilities is not 1

```python
gini_impurity_f2([0.1, 0.2, 0.3])
```

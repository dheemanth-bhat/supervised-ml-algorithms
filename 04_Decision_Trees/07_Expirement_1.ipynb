```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.tree import DecisionTreeClassifier, plot_tree
```

```python
df = pd.DataFrame(
    {
        "X1": [1, 2, 3, 4, 5, 11, 12, 13, 14, 15, 21, 22, 23, 24, 25],
        "y": ["A"] * 5 + ["B"] * 5 + ["C"] * 5,
    }
)
df
```

```python
plt.figure(figsize=(5, 4))
sns.histplot(data=df, x="X1", hue=df["y"])
plt.show()
```

```python
# Prepare our features for training
X = df[["X1"]]
y = df["y"]

clf1 = DecisionTreeClassifier(criterion="entropy", random_state=42)
clf1.fit(X, y)
```

```python
plt.figure(figsize=(8, 7))
plot_tree(clf1, feature_names=X.columns.tolist(), class_names=y.unique(), fontsize=12)
plt.show()
```

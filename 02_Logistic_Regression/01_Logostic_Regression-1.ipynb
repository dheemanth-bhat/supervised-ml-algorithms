## 2 Sigmoid function

### 2.1 Introduction

#### Formula

$$
\large
f(x)\;=\;\frac{1}{1 + \mathrm{e}^{-x}}
$$

```python
sigmoid = lambda x: 1 / (1 + np.e**-x)
```

Sigmoid is the probability of positive class.

### 2.2 Examples

#### CASE #1: Positive Distance

```python
sigmoid(4)
```

#### CASE #2: Negative Distance

```python
sigmoid(-5)
```

#### CASE #3: Zero Distance

```python
sigmoid(0)
```

## 3 Mapping Function

### 3.1 Introduction

#### Formula

$$
\newcommand{\z}{(w^T x\,+\,w_0)}
\large
\hat{y}\;=\;f(x)\;=\;\frac{1}{1 + \mathrm{e}^{-\z}}
$$

### 3.2 Examples

```python
for d in [-100, -10, -1, -0.9, -0.5, -0.001, 0, 0.001, 0.5, 0.9, 1, 10, 100]:
    y = round(sigmoid(d), 2)
    print(f"Distance: {d:>6} | Predicted prob: {y:>4}")
```

## 4 Loss Function

### 4.1 Likelihood

#### Formula

$$
\large
\text{Likelihood of a data-point}\;=\;\hat{y}^{y}\cdot\bigl(1\,-\,\hat{y}\bigr)^{1\,-\,y}
$$

```python
likelihood = lambda y, y_hat: (y_hat**y) * ((1 - y_hat) ** (1 - y))
```

#### Examples

```python
for y, y_hat in [(1, 0.9), (1, 0.2), (1, 0.5), (0, 0.1), (0, 0.7), (0, 0.5)]:
    lhood = round(likelihood(y, y_hat), 2)
    print(f"Actual y: {y} | Predicted prob: {y_hat} | Likelihood: {lhood}")
```

### 4.2 Cross-Entropy

#### Formula

$$
\large
-\sum_{i\,=\,1}^n\,\log\Bigl(\hat{y}_i^{y_i}\cdot\bigl(1\,-\,\hat{y}_i\bigr)^{1\,-\,y_i}\Bigr)
$$

- Negative Log Likelihood a.k.a Log loss a.k.a Cross-Entropy

## 5 Optimization

### 5.1 Maximum Likelihood Estimation (MLE)

#### Weight Update Rule

$$
\large
\begin{aligned}
w^{\text{new}}\;&=\;w^{\text{old}}\;-\;\frac{\eta}{n}\,\sum_{i\,=\,1}^n\,\bigl(y_i\;-\;\hat{y}_i\bigr) \cdot x_i \\[10pt]
w_0^{\text{new}}\;&=\;w_0^{\text{old}}\;-\;\frac{\eta}{n}\,\sum_{i\,=\,1}^n\,\bigl(y_i\;-\;\hat{y}_i\bigr)
\end{aligned}
$$

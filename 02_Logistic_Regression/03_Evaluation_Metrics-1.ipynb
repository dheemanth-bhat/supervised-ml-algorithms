## 2 Confusion Matrix

### 2.1 Binary Classification

Confusion Matrix has four metrics:

1. True Negative
2. False Negative
3. True Positive
4. False positive

```
                       Predicted
                   0                1
          +----------------+----------------+
        0 | True Negative  | False Positive |
Actual    +----------------+----------------+
        1 | False Negative | True Positive  |
          +----------------+----------------+
```

Above Confusion Matrix indicates the quality or **correctness of prediction**.

#### Types or Errors

```
                    Predicted
                 0              1
          +--------------+--------------+
        0 | No Error     | Type-1 Error |
Actual    +--------------+--------------+
        1 | Type-2 Error | No Error     |
          +--------------+--------------+
```

#### Accuracy using CM

$$
\newcommand{\m}[1]{\mathrm{#1}}
\large
\text{Accuracy}\;=\;\frac{\m{TP}\;+\;\m{TN}}{\m{TN}\;+\;\m{FP}\;+\;\m{TP}\;+\;\m{FN}}
$$

or

$$
\newcommand{\m}[1]{\mathrm{#1}}
\large
\text{Accuracy}\;=\;1\;-\;\frac{\m{FN}\;+\;\m{FP}}{\m{TN}\;+\;\m{FP}\;+\;\m{TP}\;+\;\m{FN}}
$$

#### Issue with Accuracy

Accuracy gives equal importance to both false positives and false negatives.

## 3 Precision

### 3.1 Introduction

#### What is Precision?

Out of all the data-points that model predicted as positive, how many of them are actually positive.

#### Formula

$$
\newcommand{\m}[1]{\mathrm{#1}}
\large
\text{Precision}\;=\;\frac{\m{TP}}{\m{TP}\;+\;\m{FP}}
$$

#### Properties

- Goal of **Precision** is to **minimize FP**, hence the emphasis is on False Positive in the denominator.
- Precision maximizes True Positive and minimizes Type-1 error.
- Precision focuses on Predicted Positives.
- Precision score range between 0 and 1.

#### Applications

- Spam detection
- Content moderation
- Predictive maintenance

## 4 Recall

### 4.1 Introduction

#### What is Recall?

Out of all the actually positive data-points, how many did the model correctly predict as positive.

#### Formula

$$
\newcommand{\m}[1]{\mathrm{#1}}
\large
\text{Recall}\;=\;\frac{\m{TP}}{\m{TP}\;+\;\m{FN}}
$$

#### Properties

- Goal of **Recall** is to **minimize FN**, hence the emphasis is on False Negative in the denominator.
- Recall maximizes True Positive and minimizes Type-2 error.
- Recall focuses on Actual Positives.
- Recall is also called as **True Positive Rate** and **Sensitivity**.
- Recall score range between 0 and 1.

#### Applications

- Cancer detection in medical domain
- Airport security
- Any critical systems in general

## 5 F1-Score

### 5.1 Introduction

#### What is F1-Score?

F1-score is harmonic mean of Precision and Recall.

##### Harmonic Mean

$$
\newcommand{\m}[1]{\mathrm{#1}}
\newcommand{\d}[1]{\displaystyle \frac{1}{\m{#1}}}
\large
\text{F1-score}\;=\;\frac{2}{\d{Precision}\;+\;\d{Recall}}
$$

#### Formula

$$
\newcommand{\m}[1]{\mathrm{#1}}
\large
\text{F1-score}\;=\;\frac{2\cdot\m{Precision}\cdot\m{Recall}}{\m{Precision}\;+\;\m{Recall}}
$$

#### Properties

- Goal of **F1-score** is to **minimize FP and FN**
- F1-score will be high only when both Precision and Recall are high.
- F1-score will be low when either one of Precision or Recall is low.
- F1-score range between 0 and 1.

#### Applications

- Fraud transaction detection in banks

````python
f1_score = (2 * 0.7 * 0.8) / (0.7 + 0.8)
round(f1_score, 2)
````

#### Odds of Success

$$
\large
\begin{aligned}
\text{Odds of Success}&\;=\;\frac{\text{Probability of success}}{\text{Probability of failure}} \\[10pt]
                      &\;=\;\frac{p(s)}{1\;-\;p(s)}
\end{aligned}
$$

```python
round(0.8 / (1 - 0.8), 2)
```

```python
round(0.1 / (1 - 0.1), 2)
```

#### How to interpret weights in Logistic Regression?

$$
\large
\text{Logistic Regression}\;=\;\text{Linear Regression of log odds}
$$

- +ve weights of $x$ has positive impact on log odds (probability of success)
- -ve weights of $x$ has negative impact on log odds (probability of success)

Then weighs can be interpreted as:

- If $x_1$ increases by one unit then log odds increases by $w_1$ units, assuming all other independent variables are constant.
- If $x_2$ increases by one unit then log odds decreases by $w_2$ units, assuming all other independent variables are constant.

# Intro to Gradient Boosting

## 1 Introduction

### 1.1 Gradient Descent

#### What is Gradient Descent?

Gradient Descent optimizes weights by minimizing some objective function or loss function.

#### How Gradient Descent works?

- Gradient Descent optimizes weights by **identifying the direction** in which the weights has to be updated.
- Gradient Descent calculates new weights as old weights **minus** the derivative of loss function w.r.t the wights.
- The impact of derivatives is controlled using Learning rate $\eta$.
- Thus optimizing the weights by minimizing a loss function.

$$
\large
w^\text{new}\;=\;w^\text{old}\;-\;\eta\,\frac{\partial\,f(w^\text{old})}{\partial\,w}
$$

where,

- $w^\text{new}$ - New updated weights
- $w^\text{old}$ - Old weights before update
- $\eta$ - Learning rate to control descent
- $f(x)$ - Any loss function

#### Why minus in weight update logic?

- Gradient Vector naturally points in the **increasing direction** of the function.
- In order to reduce or minimize the loss function, we need opposite direction of gradient.
- Hence instead of Gradient, negative gradient is used for updating the weights.

### 1.2 Functional Gradient Descent

#### Goal of Boosting

- Boosting predicts errors.
- Goal of Boosting is to predict optimal errors.

#### How Gradient Descent is used in Boosting?

- Gradient descent weight update logic updates the weights in the opposite direction of maximum.
- Boosting has no weights that needs be optimized instead Boosting **predicts errors** that needs to be minimized.
- Boosting is also interested **only in the direction** of predicted errors rather than their magnitudes.
- Hence the goal of Boosting is to reach an **optimal prediction** for every model.

$$
\large
F_{i+1}(x)\;=\;F_i(x)\;-\;\eta\,\Biggl[\frac{\partial\,L\bigl(y,\,F_i(x)\bigr)}{\partial\,F_i(x)}\Biggr]\qquad\forall\;i\in\{0,\,\dots,\,m-1\}
$$

where,

- $F_{i+1}(x)$ - New updated error predictions
- $F_i(x)$ - Old error predictions before update
- $\eta$ - Learning rate to control the impact of error prediction
- $L(y,\,F_i(x))$ - Loss function
- $m$ - Hyperparameter to control total number of base learners

#### Derivative of Loss Function

$$
\large
-\,\frac{\partial\,L\bigl(y,\,F_i(x)\bigr)}{\partial\,F_i(x)}
$$

- The derivative of loss function $L(y,\,F(x))$ w.r.t prediction $F(x)$, represents the slope of the prediction.
- Simply put, the direction in which the current prediction $F_i(x)$ should move to minimize the loss function.

#### Benefits of Gradient Descent in Boosting

1. GD enables faster convergence by reducing the number of sequential models.
2. Allows us to use any Loss Function.

### 1.3 Gradient Boosting

#### What is Gradient Boosting?

- Instead of predicting the errors Gradient Boosting predicts the direction of error.
- Gradient Boosting uses a weak learner to approximate the negative gradient.
- Gradient Boosting makes it an **Additive modeling**.

#### Additive ensemble

$$
\large
F_{i+1}(x)\;=\;F_i(x)\;+\;\eta\,h_{i+1}(x)\qquad\forall\;i\in\{0,\,\dots,\,m-1\}
$$

where,

- $F_{i+1}(x)$ - New updated error predictions
- $F_i(x)$ - Old error predictions before update
- $\eta$ - Learning rate to control the impact of error prediction
- $h(x)$ - Pseudo residual
- $m$ - Hyperparameter to control total number of base learners

#### Pseudo-residual

$$
\large
h_{i+1}(x)\;\approx\;-\,\frac{\partial\,L\bigl(y,\,F_i(x)\bigr)}{\partial\,F_i(x)}
$$

> **Note**:
>
> When solving for regression problems using MSE as loss function, the pseudo residuals is residual itself.

#### Gradient Boosting vs Functional Gradient Descent

##### How Gradient Boosting is different from Functional Gradient Descent?

- Gradient Boosting did not replace the gradient-descent equation.
- It turned the gradient-descent step into a supervised-learning problem:
  - Calculate the negative gradient, train a weak learner to approximate it, and add that learner to the existing model.
